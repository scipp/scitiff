## Intensity Channel Slicer

Scitiff image may contain 3 channel images, `intensities`, `stdevs` and `mask`.
It is because the `stdevs` or `mask` with the same size as images are needed
but cannot be stored as plain text in the metadata.

In python, the loader, `scitiff.load_scitiff` can automatically resolves the channels and place the stdevs and mask in the right place. The loaded image will only contain 1 channel after resolving with `variances` and `mask`.

However, other tiff file interfacing tools, such as ImageJ does not know about what channel means. Therefore `scitiff` has a command line tool to slice the intensity channel image and save it as a new tiff file.

```python
!scitiff-slice-channel \
    ../../tests/_regression_test_files/scitiff-26.7.0.tiff \
    --output sample_img_sliced.tiff \
    --verbose \
    --exist-ok
```

There is also a python API to slice the intensity channels from the loaded image. Command line tool `scitiff-slice-channel` uses the `scitiff.values` python API internally.

It is a wrapper of `scipp.values` function but it also handles image container(`scipp.DataGroup`) and masks.

```python
import scitiff

img = scitiff.load_scitiff('../../tests/_regression_test_files/scitiff-26.7.0.tiff', resolve_channels=False, only_image=True)
display("Image with stdevs and mask concatenated in c(channel) dimension:", img)
display("Only Intensities:", scitiff.values(img))
```
